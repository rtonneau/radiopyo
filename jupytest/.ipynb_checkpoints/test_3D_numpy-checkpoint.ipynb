# Test 3D numpy for k reactions

```python
import numpy as np
```

 k1: 2A + B ---> C + 2D
 k2:  B + C ---> D
 k3: 2C     ---> D
 ==> R1 = k1 x [A] x [A] x [B]
 ==> R2 = k2 x [B] x [C]
 ==> R3 = k3 x [C] x [C]

```python
y = np.array([5,6,7,8])
k1, k2, k3 = 10, 2, 8
```

```python
Ki = np.array([k1,k2, k3])
Riy = np.array([[-2, -1, +1, +2],[0, -1, -1, +1], [0,0,-2,1]]).T
Rki = np.zeros([4,3,3])

#First reaction
Rki[0,0,0] = 1
Rki[0,0,1] = 1
Rki[1,0,2] = 1

#Second reaction
Rki[1,1,0] = 1
Rki[2,1,1] = 1
Rki[:,1,1:] = np.inf

#Third reaction
Rki[2,2,0] = 1
Rki[2,2,1] = 1
Rki[:,2,1:] = np.inf

res = np.einsum("j, jik -> ik ", y, Rki[:,:,:])
```
